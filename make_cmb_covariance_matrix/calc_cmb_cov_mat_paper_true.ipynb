### 40GHz_beam w/o wl

```julia
# truncate covariance matrix 
include("smoothing_cov_mat_calc_bandpass_40GHz_beam_wo_wl.jl")
# cmb_map3に対応した 共分散行列の計算をする
using NPZ

cl_2048 = npzread("planck_cl_th_nside_2048.npz")
nside = 8
lmin = 2
#lmax = 8
lmax = 2nside
cl_scal = cl_2048["clL_th"]
cl_tens = cl_2048["clt_th"];

# ここミスってた、 lensed scalarは scalのlenseing込みの状態なので、unlensed scalar + lensingはダブルカウントになってた
cl_ee_scal = cl_scal[:, 2]
cl_ee_tens = cl_tens[:, 2]
cl_bb_scal = cl_scal[:, 3]
cl_bb_tens = cl_tens[:, 3];
cov_mat_scal = calc_cmb_cov_mat(cl_ee_scal, cl_bb_scal, nside, lmin, lmax)
cov_mat_tens = calc_cmb_cov_mat(cl_ee_tens, cl_bb_tens, nside, lmin, lmax)

npzwrite("../paper_smoothing_cov_mat_scal_nside_$(nside)_lmin_$(lmin)_lmax_$(lmax)_40GHz_beam_wo_wl.npy", cov_mat_scal)
npzwrite("../paper_smoothing_cov_mat_tens_nside_$(nside)_lmin_$(lmin)_lmax_$(lmax)_40GHz_beam_wo_wl.npy", cov_mat_tens)
```
