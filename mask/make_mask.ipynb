### make mask

```julia
function threshold_to_binary(matrix::AbstractArray{T}, threshold::Float64) where T
    # threshold以下の要素を0.0、thresholdより大きい場合を1.0に変換
    return map(x -> x <= threshold ? 0.0 : 1.0, matrix)
end
```

```julia
mask_512 = hp.read_map("../../wmap_polar_r9_mask_3yr_v2.fits", field=1);
```

```julia
function save_mask_with_nside(mask::Vector{Float32}, nside::Int, threshold::Float64, output_dir::String)
    mask_rescaled = threshold_to_binary(hp.ud_grade(mask, nside), threshold)
    output_file = joinpath(output_dir, "P06_nside_$nside.fits")
    hp.write_map(output_file, mask_rescaled, overwrite=true)
    println("Saved mask with nside=$nside to $output_file")
end
```

```julia
save_mask_with_nside(mask_512, 32, 0.98, "../../")
```
